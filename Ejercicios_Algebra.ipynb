```


import sympy as sp
import geopandas as gpd

print("===== T1: SISTEMAS DE ECUACIONES =====")


print("\n--- Solución única ---")

A1 = sp.Matrix([[3, 2, -1],
                [2, -2, 4],
                [-1, 0.5, -1]])
b1 = sp.Matrix([1, -2, 0])

print("Ejercicio 1:", A1.LUsolve(b1))

A2 = sp.Matrix([[2, 1, 1],
                [1, 3, 2],
                [1, 0, 0]])
b2 = sp.Matrix([4, 5, 6])

print("Ejercicio 2:", A2.LUsolve(b2))



print("\n--- Infinitas soluciones ---")

A3 = sp.Matrix([[1, 1, 1],
                [2, 2, 2],
                [3, 3, 3]])
b3 = sp.Matrix([2, 4, 6])

print("Ejercicio 1:", A3.gauss_jordan_solve(b3))

A4 = sp.Matrix([[2, -2, 1],
                [4, -4, 2],
                [6, -6, 3]])
b4 = sp.Matrix([1, 2, 3])

print("Ejercicio 2:", A4.gauss_jordan_solve(b4))



print("\n--- Sistema inconsistente ---")

A5 = sp.Matrix([[1, -1, 2],
                [2, -2, 4],
                [3, -3, 6]])
b5 = sp.Matrix([1, 2, 5])

try:
    print("Ejercicio 1:", A5.gauss_jordan_solve(b5))
except:
    print("Ejercicio 1: No tiene solución")

A6 = sp.Matrix([[2, 3, 1],
                [4, 6, 2],
                [1, 1, 0]])
b6 = sp.Matrix([5, 10, 3])

try:
    print("Ejercicio 2:", A6.gauss_jordan_solve(b6))
except:
    print("Ejercicio 2: No tiene solución")




print("\n===== T2: DETERMINANTES =====")

M = sp.Matrix([[1, 2, 3],
               [0, 4, 5],
               [1, 0, 6]])

# Gauss
print("\nGauss:", M.det())

# Diagonalización
eigs = M.eigenvals()
det_diag = 1
for val in eigs:
    det_diag *= val

print("Diagonalización:", det_diag)

# Cofactores
print("Cofactores:", M.det(method='berkowitz'))




print("\n===== T3: PROPIEDADES =====")

A = sp.Matrix([[2,3,1],
               [4,1,5],
               [7,2,6]])

# 1
B = A.copy()
B.row_swap(1,2)
print("Propiedad 1:", A.det(), B.det())

# 2
C = A.copy()
C[1,:] = 3*C[1,:]
print("Propiedad 2:", A.det(), C.det())

# 3
D = sp.Matrix([[2,3,1],
               [2,3,1],
               [7,8,9]])
print("Propiedad 3:", D.det())

# 4
E = sp.Matrix([[0,0,0],
               [2,3,4],
               [5,6,7]])
print("Propiedad 4:", E.det())

# 5
F = A.copy()
F[2,:] = F[2,:] + F[0,:]
print("Propiedad 5:", A.det(), F.det())

# 6
G = sp.Matrix([[3,1,2],
               [0,5,4],
               [0,0,7]])
print("Propiedad 6:", G.det())

# 7
I = sp.eye(3)
print("Propiedad 7:", I.det())

# 8
H = sp.Matrix([[2,0,1],
               [1,3,0],
               [4,1,2]])

J = sp.Matrix([[1,2,0],
               [0,1,3],
               [2,0,1]])

print("Propiedad 8:", (H*J).det(), H.det()*J.det())

# 9
print("Propiedad 9:", H.det(), H.T.det())




print("\n===== EXTRA: GeoPandas =====")

data = {
    'Ciudad': ['X', 'Y', 'Z'],
    'Latitud': [10.0, 10.5, 11.0],
    'Longitud': [-75.0, -75.5, -76.0]
}

geo_df = gpd.GeoDataFrame(
    data,
    geometry=gpd.points_from_xy(data['Longitud'], data['Latitud'])
)

print(geo_df)

print("\n=== FIN ===")
```
